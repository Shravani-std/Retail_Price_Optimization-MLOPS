```python
class DataProcessing:
  def __init__(self, df = pd.DataFrame):
    self.df = df

  def check_missing_values(self) -> pd.Series:
    missing_values = self.df.isnull().sum()
    return missing_values

  def check_duplicate_values(self):
    duplicate_values = self.df.duplicated().sum()
    return duplicate_values

  def check_dataType(self):
    data_types = self.df.dtypes
    return data_types
```

```python
  import matplotlib.pyplot as plt
  import seaborn as sns
```

```python
correlation_matrix = df.select_dtypes(include=[np.number]).corr(method = 'pearson')
correlation_matrix
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm')
plt.title("Correlation Matrix HeatMap")
plt.show()
```

```python

class CorrelationAnalysis:

  def __init__(self, df: pd.DataFrame):

      self.df = df.select_dtypes(include=[np.number])
      self.correlation_matrix = None

  def calculate_correlation_matrix(self,method = 'pearson') -> pd.DataFrame:
    self.correlation_matrix = self.df.corr(method=method)
    return self.correlation_matrix

  def plot_correlation_heatMap_matrix(self):
    if self.correlation_matrix is None:
      raise ValueError("Correlation Matrix is not calculated")
    else:
      plt.figure(figsize=(10,10))
      sns.heatmap(self.correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm')
      plt.title("Correlation Matrix HeatMap")
      plt.show()

matrix = CorrelationAnalysis(df)
matrix.calculate_correlation_matrix()
matrix.plot_correlation_heatMap_matrix()
```


# **📌 Correlation Analysis**
Correlation analysis helps us understand relationships between variables in the dataset. Key insights:

🔗 1. Quantity (qty) vs Total Price (total_price)

Observation: Strong positive correlation.

Interpretation: As the quantity sold increases, the total price also increases.

Reasoning: This is expected since total_price is directly dependent on qty.

🔗 2. Unit Price (unit_price) vs Competitor Prices (comp_1, comp_2, comp_3)

Observation: Likely a positive correlation.

Interpretation: Our product prices tend to move in line with competitors.

Reasoning: Competitive pricing strategies often result in similar price patterns across brands.

🔗 3. Product Weight (product_weight_g) vs Freight Price (freight_price)

Observation: Positive correlation.

Interpretation: Heavier products generally have higher shipping costs.

Reasoning: Freight charges increase with product weight, making this relationship logical.

🔗 4. Product Score (product_score) vs Customers (customers)

Observation: Potential positive correlation.

Interpretation: Products with higher ratings attract more customers.

Reasoning: Higher customer satisfaction (better scores) may lead to increased demand.
